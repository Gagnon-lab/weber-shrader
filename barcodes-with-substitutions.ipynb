```python
import numpy as np
import pandas as pd
```

```python
experiment = '07.31.2025_01.42_PM'
file = f'./experiments/{experiment}/final_data.csv'
```

```python
data = pd.read_csv(file)
```

```python
def has_substitution(barcode):
    return 'S' in barcode
```

```python
data['has_substitution'] = data['barcode'].apply(has_substitution)
```

```python
data[data['has_substitution']]
```

```python
data[data['has_substitution']].groupby(['barcode', 'fish'], as_index = False).aggregate({'reads': 'sum'})
```

```python
np.array(data[data['has_substitution']].groupby(['barcode', 'fish'], as_index = False).aggregate({'reads': 'sum'})['barcode'])
```
