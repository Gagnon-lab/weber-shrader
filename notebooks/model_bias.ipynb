```python
import numpy as np
import pandas as pd

from analysis import *
```

```python
rows = []

for rp in [0, 0.5, 2, 5, 8, 15, 25]:
    df2 = df.query(f'scenario == "r36_rp{rp}_mimic"')

    estimate = df2['r_opt']
    lower = df2['lower']
    upper = df2['upper']

    mean_est = estimate.mean()
    sd_est   = estimate.std()

    if rp == 0:
        rel_bias = np.nan
    else:
        rel_bias = (mean_est - rp) / rp * 100

    ci_contains = ((lower <= rp) & (rp <= upper)).mean() * 100

    rows.append({
        'True rp': rp,
        'Mean ± SD estimate': f'{mean_est:.2f} ± {sd_est:.2f}',
        'Relative bias (%)': f'{rel_bias:.2f}%',
        '% CI containing true rp': f'{ci_contains:.2f}%'
    })

result_table = pd.DataFrame(rows)
result_table
```
